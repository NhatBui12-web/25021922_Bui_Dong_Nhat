```python

#W9A1
def merge_recursive(a,b):
  if not a:
    return b
  if not b:
    return a
  res = []
  if a[0]>b[0]:
    res = [b[0]] + merge_recursive(a,b[1:])
  else:
    res = [a[0]] + merge_recursive(a[1:],b)
  return res
a= list(map(int,input().split()))
b= list(map(int,input().split()))
print(*merge_recursive(a,b))
```

```python

#W9A2
def findk(a):
  res = []
  if not a:
    return []
  else:
    min_val = min(a)
    a.remove(min(a))
    res = [min_val] + findk(a)
  return res
a = []
n = int(input())
for i in range(n):
  a.extend(list(map(int,input().split())))
k = int(input())
res = findk(a)
print(res[k-1])
```

```python
#w9a3
def s(n):
     a=list(n)
     n_min=min(a,key=lambda x:int(x))
     if len(a)==2:
          return a[1]+a[0]
     elif a[0]==n_min:
          return a[0]+s("".join(a[1::]))
     else:
          for i in range(len(a)-1,-1,-1):
               if a[i]==n_min:
                    a[0],a[i]=a[i],a[0]
                    return "".join(a) 
n=input()
print(int(s(n)))
```

```python
#w9a4
t = int(input())
for _ in range(t):
    a = input()
    lst_a = list(a)
    b = input()
    check = True
    tmp = list(a)
    for ch in lst_a:
        if ch.upper() not in b:
            tmp.remove(ch)
    if len(tmp) != len(b):
        check = False
    else:
        a = "".join(tmp)
        if a.upper() != b:
          check = False
    if check:
        print('YES')
    else:
        print('NO')
```

```python

#W9A6
n = int(input())
a = list(map(int,input().split()))
i = n - 2
while i >= 0 and a[i] >= a[i+1]:
    i -= 1
if i == -1:
    a.reverse()
else:
    j = n-1
    while a[j] <= a[i]:
        j -= 1
    a[i],a[j] = a[j],a[i]
    a[i+1:] = reversed(a[i+1:])
print(*a)
```

```python
#w9a7
def canPlaceFlowers(flowerbed, k):
     so_vi_tri=0
     if flowerbed==[1]:return False
     elif flowerbed==[0]:
          if k<=1:return True
          else:return False
     else:
          for i in range(len(flowerbed)):
               if i==0:
                    if flowerbed[0]==flowerbed[1]==0:
                         so_vi_tri+=1
                         flowerbed[0]=1
               elif i==len(flowerbed)-1:
                    if flowerbed[len(flowerbed)-1]==flowerbed[len(flowerbed)-2]==0:
                         so_vi_tri+=1
                         flowerbed[len(flowerbed)-1]=1
               else:
                    if flowerbed[i-1]==flowerbed[i]==flowerbed[i+1]==0:
                         flowerbed[i]=1
                         so_vi_tri+=1
     return so_vi_tri>=k
flowerbed=list(map(int,input().split()))
k=int(input())
print(canPlaceFlowers(flowerbed, k))
```

```python
#w9a8
def is_path_crossing(moves):
     x,y=0,0
     duong_di={(0,0)}
     for i in moves:
          if i == "R": x += 1
          elif i == "L": x -= 1
          elif i == "U": y += 1
          else: y -= 1
     
          if (x, y) in duong_di:
               return True
          duong_di.add((x, y))
     return False
moves=input()
print(is_path_crossing(moves))
```

```python

#W9A9
def Median(nums):
    a,b,c,d,e = nums
    if a > b:
        a,b=b,a
    if c>d:
        c,d=d,c
    if a>c:
        a,c=c,a
        b,d=d,b
    if e < b:
        return b
    if e > c:
        return c
    return e
nums = list(map(int,input().split()))
print(Median(nums))
```

```python

#W9A10
def findRadius(houses,heaters):
    houses.sort()
    heaters.sort()
    i = 0
    r = 0
    for h in houses:
        while i < len(heaters) -1 and abs(heaters[i+1]-h) <= abs(heaters[i]-h):
            i+=1
        r = max(r,abs(heaters[i] - h))
    return r
houses = list(map(int,input().split()))
heaters = list(map(int,input().split()))
print(findRadius(houses,heaters))
```
